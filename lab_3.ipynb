```python
import pandas as pd

def years_active(df):
    indices = df['Plant ID']         
    liste = []
    
    for i in range(2020, 2031):       
        li = []
        for el in indices:
          
            re_df = df[df['Plant ID'] == el].iloc[0]

            start_str = str(re_df['Start date_x'])
            start_year = int(start_str[:4])   

         
            retired_val = re_df['Retired date']
            retired_str = str(retired_val)

            if not retired_str[:4].isdigit():
                retired_year = 2030
            else:
                retired_year = int(retired_str[:4])

        
            if start_year <= i <= retired_year:
                li.append(1)
            else:
                li.append(0)

        liste.append(li)

    j = 2020
    while j <= 2030:
        df[str(j)] = liste[j - 2020]
        j += 1                          

    return df

china = years_active(china)
```

```python
# compute country-level and technology-level steel utilization rates
# uses existing dataframes: plant_capacities, plant_production (and merged_df if desired)
# depends on pandas (imported in earlier cells) and numpy (imported earlier as np)

def _find_col(df, candidates):
    for c in candidates:
        if c in df.columns:
            return c
    return None

def _parse_num_series(s):
    # remove non-numeric characters except dot and minus, then coerce
    return pd.to_numeric(s.astype(str).str.replace(r'[^0-9.\-]', '', regex=True), errors='coerce')

def compute_country_utilization(country='China', years=range(2019, 2025), by_tech=True,
                                df_cap=plant_capacities, df_prod=plant_production):
    # locate relevant column names
    cap_country_col = _find_col(df_cap, ['Country/Area', 'Country/Area_x', 'Country/Area_y'])
    cap_pid_col = _find_col(df_cap, ['Plant ID'])
    prod_pid_col = _find_col(df_prod, ['Plant ID'])
    year_cols = [str(y) for y in years]
    # filter capacity rows for country
    cap_df = df_cap[df_cap[cap_country_col] == country].copy()
    if cap_df.empty:
        raise ValueError(f'No capacity rows found for country {country}')
    # clean numeric capacity columns
    cap_df['cap_total'] = _parse_num_series(cap_df.get('Nominal crude steel capacity (ttpa)', 
                                                       cap_df.get('Nominal crude steel capacity', pd.Series())))
    cap_df['cap_bof'] = _parse_num_series(cap_df.get('Nominal BOF steel capacity (ttpa)', 
                                                     cap_df.get('Nominal BOF steel capacity', pd.Series())))
    cap_df['cap_eaf'] = _parse_num_series(cap_df.get('Nominal EAF steel capacity (ttpa)', 
                                                     cap_df.get('Nominal EAF steel capacity', pd.Series())))
    # aggregate capacity by Plant ID (some datasets have duplicates)
    cap_agg = cap_df.groupby(cap_pid_col)[['cap_total','cap_bof','cap_eaf']].sum(min_count=1).reset_index()
    # prepare production: parse numeric and aggregate by Plant ID for requested years
    prod = df_prod[[prod_pid_col] + [c for c in year_cols if c in df_prod.columns]].copy()
    for c in year_cols:
        if c in prod.columns:
            prod[c] = _parse_num_series(prod[c])
        else:
            prod[c] = np.nan
    prod_agg = prod.groupby(prod_pid_col)[year_cols].sum(min_count=1).reset_index()
    # merge capacities and production
    merged = pd.merge(cap_agg, prod_agg, left_on=cap_pid_col, right_on=prod_pid_col, how='left')
    merged.fillna({c:0 for c in ['cap_total','cap_bof','cap_eaf']}, inplace=True)
    results = []
    for y in year_cols:
        total_prod = merged[y].sum(min_count=1)
        total_cap = merged['cap_total'].sum(min_count=1)
        util = np.nan if (pd.isna(total_cap) or total_cap == 0) else total_prod / total_cap
        row = {'year': int(y), 'production': total_prod, 'capacity': total_cap, 'utilization': util}
        if by_tech:
            # allocate each plant's production to BOF/EAF proportionally to that plant's BOF/EAF capacity
            bo_prod = 0.0
            ea_prod = 0.0
            for _, r in merged.iterrows():
                p = r[y] if not pd.isna(r[y]) else 0.0
                bo = r['cap_bof'] if not pd.isna(r['cap_bof']) else 0.0
                ea = r['cap_eaf'] if not pd.isna(r['cap_eaf']) else 0.0
                tech_sum = bo + ea
                if tech_sum > 0:
                    bo_prod += p * (bo / tech_sum)
                    ea_prod += p * (ea / tech_sum)
                else:
                    # if plant lacks tech breakdown, assign production proportionally to national tech capacity share
                    # compute later after loop
                    pass
            # handle plants without tech breakdown by distributing their production according to national tech shares
            no_tech_mask = (merged['cap_bof'] + merged['cap_eaf'] == 0)
            no_tech_prod = merged.loc[no_tech_mask, y].sum(min_count=1)
            total_bof_cap = merged['cap_bof'].sum(min_count=1)
            total_eaf_cap = merged['cap_eaf'].sum(min_count=1)
            tech_cap_sum = (total_bof_cap if not pd.isna(total_bof_cap) else 0) + (total_eaf_cap if not pd.isna(total_eaf_cap) else 0)
            if tech_cap_sum > 0 and not pd.isna(no_tech_prod) and no_tech_prod > 0:
                bo_prod += no_tech_prod * (total_bof_cap / tech_cap_sum)
                ea_prod += no_tech_prod * (total_eaf_cap / tech_cap_sum)
            bo_util = np.nan if (total_bof_cap == 0 or pd.isna(total_bof_cap)) else bo_prod / total_bof_cap
            ea_util = np.nan if (total_eaf_cap == 0 or pd.isna(total_eaf_cap)) else ea_prod / total_eaf_cap
            row.update({
                'production_bof_est': bo_prod,
                'production_eaf_est': ea_prod,
                'capacity_bof': total_bof_cap,
                'capacity_eaf': total_eaf_cap,
                'utilization_bof': bo_util,
                'utilization_eaf': ea_util
            })
        results.append(row)
    return pd.DataFrame(results).set_index('year')

# Example: compute China utilization for 2019-2024 and show 2023
china_util = compute_country_utilization(country='China', years=range(2019,2025), by_tech=True)
print(china_util)
print("\nChina utilization 2023:", china_util.loc[2023].to_dict())
```
